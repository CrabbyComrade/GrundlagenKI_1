Perzeptron.01
    1.
        Trennebene E: 2x1 ​+x2​ +x3 ​= 0
        Klassifikation +1, wenn 2x1 ​+x2​ +x3 >0
    2.
        Perzeptron 1 und 2 haben dieselbe Trennebene und Klassifikation (Da in beiden x1 doppelt so viel Gewicht hat wie x2 und x3)
        Perzeptron 4 hat dieselbe Trenngerade wie 1 und 2, aber umgekehrte Klassifikation, da alle Gewichtungen negativ sind
        Perzeptron 3 hat weder gemeinsame Trenngerade noch Klassifikation mit den anderen. (sqrt(2) ist nicht 2*sqrt(2), also hat x1 nicht doppelt so viel Gewicht wie x2, x3)
        

```python
# Perzeptron.02.1
def perceptron(weights, values):
    result = 0
    for i in range(len(weights)):
        weight = weights[i]
        if i == len(weights)-1:
            result += weight                # Last Weight is bias
        else:
            result += weight * values[i]    # weighting of x1, x2 etc.
    # Classification
    if result>0:
        return +1
    elif result<0:
        return -1
    else:
        return 0

# AND
weights = [1, 1, -1.5]
values_true = [1, 1]
values_false = [0, 1]
print(perceptron(weights, values_true))
print(perceptron(weights, values_false))

# OR
weights = [1, 1, -0.5]
values_true = [1, 0]
values_false = [0, 0]
print(perceptron(weights, values_true))
print(perceptron(weights, values_false))

# COMPLEMENT / NEGATION 
weights = [-1.5, 1]
values_true = [0]
values_false = [1]
print(perceptron(weights, values_true))
print(perceptron(weights, values_false))
```

```python
# Perzeptron.03.02
import random

def perceptron(weights, values):
    result = 0
    for i in range(len(weights)):
        weight = weights[i]
        if i == len(weights)-1:
            result += weight                # Last Weight is bias
        else:
            result += weight * values[i]    # weighting of x1, x2 etc.
    # Classification
    if result>0:
        return +1
    elif result<0:
        return -1
    else:
        return 0

data = [(False, (random.uniform(-1,1),random.uniform(-1,1))) for _ in range(10)] # False/True for Classification -1/+1
line = (random.uniform(-1,1),random.uniform(-1,1),random.uniform(-1,1))
# Classifies dataset using the correct line/weights
for i in range(len(data)):
    if perceptron(line, data[i][1]) > 0:
        data[i] = (True, data[i][1])


def perceptron_training(weights, data, maxSteps, learning_rate):
    for step in range(maxSteps):
        mistakes = []

        # Find misclassifications
        for d in data:
            result = perceptron(weights, d[1])  # Get classification with current weights
            if result == 1: result = True       # Put True for +1
            else: result = False                # Put False for -1 (and 0 but that's negligible)
            if result != d[0]:                  # Compare to correct result
                mistakes.append(d)              # Add to mistakes if applicable
        
        if len(mistakes) == 0:
            return (weights, step)
        
        # Chose random mistake and get current(wrong) classification
        mistake = random.choice(mistakes)
        current_classification = perceptron(weights, mistake[1])

        # Calculate error (0 for no error or -2/+2)
        if mistake[0] == True:
            error = 1 - current_classification
        else:
            error = -1 -current_classification
        
        # Update weights and bias according to: w := w + α ( y ( i ) − h ( x ( i ) ) ) x ( i )
        updated_weights = list(weights)
        for i in range(len(mistake[1])):
            updated_weights[i] += learning_rate * error * mistake[1][i] # weights
        updated_weights[-1] += learning_rate * error                    # bias
        weights = tuple(updated_weights)
    return (weights, maxSteps)

start_weights = (0,0,0)
highest_steps = -1
lowest_steps = -1
total_steps = 0
for _ in range(1000):
    value = perceptron_training(start_weights, data, 1000, 1)[1]
    if(value > highest_steps):
        highest_steps = value
    if(value < lowest_steps or lowest_steps==-1):
        lowest_steps = value
    total_steps += value
print(f"Trained 1000x, average: {total_steps/1000}, lowest necessary steps : {lowest_steps}, highest necessary steps : {highest_steps}")
```
